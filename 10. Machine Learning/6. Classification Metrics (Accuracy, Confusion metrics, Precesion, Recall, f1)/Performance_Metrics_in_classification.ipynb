```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score
)

# Example true labels and predictions
y_true = [0, 1, 1, 0, 1, 0, 1, 1]
y_pred = [0, 1, 0, 0, 1, 0, 1, 0]

# Accuracy
accuracy = accuracy_score(y_true, y_pred)
print("Accuracy:", accuracy)

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:\n", cm)

# Precision
precision = precision_score(y_true, y_pred)
print("Precision:", precision)

# Recall
recall = recall_score(y_true, y_pred)
print("Recall:", recall)

# F1 Score
f1 = f1_score(y_true, y_pred)
print("F1 Score:", f1)

# F-beta Score (example: beta = 0.5 gives more weight to precision)
fbeta = fbeta_score(y_true, y_pred, beta=0.5)
print("F-beta Score (beta=0.5):", fbeta)
```

# Focus on Precision when False Positives are very costly. This means you want to be very sure about your positive predictions. (e.g., Medical diagnosis for a rare, serious disease where false alarms cause unnecessary distress/treatment).

Focus on Recall when False Negatives are very costly. This means you want to catch as many actual positive cases as possible. (e.g., Fraud detection, where missing a fraudulent transaction is worse than a few false alarms).

F-beta Score helps you combine Precision and Recall, with a customizable weight:

beta = 1: Gives equal importance to Precision and Recall (this is the F1-score).
beta > 1 (e.g., beta = 2): Prioritizes Recall. Use this when False Negatives are more critical than False Positives. A higher beta value means you want to avoid missing true positives even if it means more false alarms.
beta < 1 (e.g., beta = 0.5): Prioritizes Precision. Use this when False Positives are more critical than False Negatives. A lower beta value means you want to avoid false alarms, even if it means missing some true positives.
Choose the beta value based on your problem's specific business context and the relative costs of False Positives versus False Negatives. If missing a positive (FN) is twice as bad as a wrong positive (FP), use beta=2. If a wrong positive (FP) is twice as bad as missing a positive (FN), use beta=0.5.
