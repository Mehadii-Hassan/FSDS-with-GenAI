```python
import pandas as pd
import seaborn as sns
```

```python
titanic = pd.read_csv("/content/train.csv")
```

```python
titanic.head()
```

Load tips dataset

```python
tips = sns.load_dataset("tips")
tips.head()
```

Load flight dataset

```python
flights = sns.load_dataset("flights")
flights.head()
```

Load iris dataset

```python
iris = sns.load_dataset("iris")
iris.head()
```

# Scatterplot (Numerical - Numerical)

```python
tips.head()
```

Relation with "total_bill" and "tip"

```python
sns.scatterplot(data = tips, x = "total_bill", y = "tip")
```

```python
sns.scatterplot(data = tips, x = "total_bill", y = "tip", hue = tips["sex"]) #check male & female
```

```python
sns.scatterplot(data = tips, x = "total_bill", y = "tip", hue = tips["sex"], style = tips["smoker"]) #check male & female
```

# Bar plot (Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(data = titanic, x = "Pclass", y = "Age")
```

```python
sns.barplot(data = titanic, x = "Pclass", y = "Age", hue = titanic["Sex"]) #Check male & female
```

# Box plot (Numerical - Categorical)

```python
sns.boxplot(data = titanic, x = "Pclass", y = "Age")
```

```python
sns.boxplot(data = titanic, x = "Pclass", y = "Age", hue = titanic["Survived"])
```

# Distplot (Numerical - Categorical)

```python
sns.distplot(titanic[titanic["Survived"]==0]["Age"], hist=False)
```

```python
sns.distplot(titanic[titanic["Survived"]==1]["Age"], hist=False)
```

```python
sns.distplot(titanic[titanic["Survived"]==0]["Age"], hist=False)
sns.distplot(titanic[titanic["Survived"]==1]["Age"], hist=False)
```
